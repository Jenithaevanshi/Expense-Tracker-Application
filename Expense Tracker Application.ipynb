```python
transactions = []
```

```python
def add_income():
    date = input("Enter the Date DD-MM-YYYY :")
    category = input("Enter Income Category:")
    amount= float(input("Enter Income Amount:"))
    description = input("Enter Description:")

    income = {
        "Type": "Income",
        "Date": date,
        "Category": category,
        "Amount": amount,
        "Description":description
    }

    transactions.append(income)
    print("\n Income successfully added")
    
    
```

```python
add_income()
```

```python
def add_expense():
    date = input("Enter the Date DD-MM-YYYY:")
    category = input("Enter the Category:")
    amount = float(input("Enter the Amount:"))
    description = input("Enter description:")

    expense = {
        "Type": "Expense",
        "Date": date,
        "Category": category,
        "Amount": amount,
        "Description": description
    }

    transactions.append(expense)
    print("\n Expense successfully added")
```

```python
add_expense()
```

```python
def display_transactions():
    if len(transactions) == 0:
        print("\n No Transactions found:")
        return
    print("\n ----All transactions----")
    for transaction in transactions:
        
        print("----------------")
        print("Type:",transaction["Type"])
        print("Date:",transaction["Date"])
        print("Category:",transaction["Category"])
        print("Amount:",transaction["Amount"])
        print("Description:",transaction["Description"])
        
```

```python
display_transactions()
```

```python
import csv
transactions = []
```

```python
def load_from_csv():
    transactions.clear()
    with open("expense.csv","r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            row["Amount"]=float(row["Amount"])
            transactions.append(row)
    print("Transaction loaded successfully")
            
        
```

```python
load_from_csv()
print(transactions)
```

```python
def search_category():
    category = input("Enter category")
    found = False
    for transaction in transactions:
        if transaction["Category"].lower() == category.lower():
            print(transaction)
            found = True
    if not found:
        print("No records found.")
```

```python
def generate_report():

    if len(transactions) == 0:
        print("No Transactions Found.")
        return

    total_income = 0
    total_expense = 0

    category_summary = {}

    highest_expense = None

    monthly_income = {}
    monthly_expense = {}

    for transaction in transactions:

        # Income
        
        if transaction["Type"] == "Income":

            total_income += transaction["Amount"]

            # Monthly Income
            month = transaction["Date"][3:10] # 08-08-2026

            if month in monthly_income:
                monthly_income[month] += transaction["Amount"]
            else:
                monthly_income[month] = transaction["Amount"]


        
        # Expense
       
        else:

            total_expense += transaction["Amount"] # total_expense= total_expense+transaction["Amount"]

            # Highest Expense
            if highest_expense is None:
                highest_expense = transaction

            elif transaction["Amount"] > highest_expense["Amount"]:
                highest_expense = transaction


            # Category Summary
            category = transaction["Category"]

            if category in category_summary:
                category_summary[category] += transaction["Amount"]
            else:
                category_summary[category] = transaction["Amount"]


            # Monthly Expense
            month = transaction["Date"][3:10]

            if month in monthly_expense:
                monthly_expense[month] += transaction["Amount"]
            else:
                monthly_expense[month] = transaction["Amount"]

    # Balance
   
    balance = total_income - total_expense

    # Main Report

    print("\n========== EXPENSE REPORT ==========")

    print("Total Income :", total_income)
    print("Total Expense :", total_expense)
    print("Balance :", balance)

    # Category Summary
  
    print("\n------ Category-wise Expenses ------")

    for category, amount in category_summary.items():

        print(category, ":", amount)



    # Monthly Income
   

    print("\n------ Monthly Income ------")

    for month, amount in monthly_income.items():

        print(month, ":", amount)


  
    # Monthly Expense
   

    print("\n------ Monthly Expense ------")

    for month, amount in monthly_expense.items():

        print(month, ":", amount)


  
    # Highest Expense


    if highest_expense is not None:

        print("\n------ Highest Expense ------")

        print("Category :", highest_expense["Category"])
        print("Amount :", highest_expense["Amount"])
        print("Date :", highest_expense["Date"])

# Main Menu

while True:

    print("\n===================================")
    print(" Expense Tracker Application ")
    print("===================================")

    print("1. Add Income")
    print("2. Add Expense")
    print("3. Display Transactions")
    print("4. Search Category")
    print("5. Save to CSV")
    print("6. Load from CSV")
  
    print("7. Generate Report")
    print("8. Exit")

    choice = input("Enter Your Choice : ")

    if choice == "1":
        add_income()

    elif choice == "2":
        add_expense()

    elif choice == "3":
        display_transactions()

    elif choice == "4":
        search_category()

    elif choice == "5":
        save_to_csv()

    elif choice == "6":
        load_from_csv()

    elif choice == "7":
        generate_report()

    elif choice == "8":
        print("Thank You!")
        break

    else:
        print("Invalid Choice.")
```
